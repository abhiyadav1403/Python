```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 6,9)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("chart")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

```python
# Subplot in matplot

plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x,y)
```

Object Oriented way

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).


```python
fig = plt.figure(figsize=(20,8), dpi = 100)
axis1 = fig.add_axes([0.1,0.1,0.4,0.7])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(x,y)
```
