```python
import sqlite3
```

```python
# connect to an   SQLite database
connection= sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
# create a Table
cursor.execute( '''
Create Table If Not Exists employees(
        id Integer Primary Key,
        name Text Not Null,
        age Integer,
        department Text 
        )
''')


# commit the changes

connection.commit()
```

```python
cursor.execute(''' 
Select * from employees
''')
```

```python
# Insert data in sqlite table
cursor.execute('''
Insert Into employees(name, age, department) 
values("Abhishek", 23, "Data Analyst")
''')
cursor.execute('''
Insert Into employees(name, age, department) 
values("Abhi", 21, "Data Science")
''')
cursor.execute('''
Insert Into employees(name, age, department) 
values("Shubham", 20, "Web Development")
''')
cursor.execute('''
Insert Into employees(name, age, department) 
values("Jyoti", 21, "MERN Stack")
''')
cursor.execute('''
Insert Into employees(name, age, department) 
values("Rohit", 24, "Full Stack")
''')

# commit the changes
connection.commit()
```

```python
# Query the data from the table
cursor.execute('''
select * from employees''')

rows = cursor.fetchall()

#  print the queried data
for row in rows:
    print(row)
```

```python
# working with Sales Data
connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

# create a table for sales data
cursor.execute('''
create table if not exists sales(
id integer primary key,
date text not null,
product text not null,
sales integer,
region text
)

''')
sales_data = [
    ('2024-01-01', 'Product A', 100, 'North'),
    ('2024-01-02', 'Product B', 150, 'South'),
    ('2024-01-03', 'Product C', 200, 'East'),
    ('2024-01-04', 'Product D', 250, 'West'),
]

cursor.executemany('''
    insert into sales(date , product , sales, region) values(?,?,?,?)
''', sales_data)

connection.commit()
```

```python
# Query the data from the table
cursor.execute('select * from sales')
rows = cursor.fetchall()

# print the queried data
for row in rows:
    print(row)
```

```python
# close the connection
connection.close()
```
